```python
import numpy as pd
import pandas as pd
```

```python
data = pd.read_csv(r'anime.csv')
```

```python
df = pd.DataFrame(data)
```

```python
#feature extraction
#no seperate col. but we will extract the episode count from title
def extract_episode(txt):
  data = ""
  check = False
  for i in txt:
    if i == '(':
      check = True
      continue
    if i == ')':
      break
    if check == True:
      data += i
  return data if data else None
```

```python
#apply the function on Title
df['Episodes'] = df['Title'].apply(extract_episode)
```

```python
#Now remove the _eps from the epsidoes last
df['Episodes'] = df['Episodes'].str.replace(' eps', '')
```

```python
#Now convert in into the integer
df['Episodes'] = df['Episodes'].astype(int)
```

```python
#Now extracting timestamp from title again defining func
def extract_timestamp(txt):
  count = 0
  for i in txt:
    count+=1
    if i == ')':
      break
  data = ""
  for i in range(count,count+19):
    data+=txt[i]
  return data
```

```python
#apply the feature and create a new col Time_Stamp
df['TimeStamp'] = df['Title'].apply(extract_timestamp)
```

```python
#Now calculate total months used chatgpt to define the functions
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calculate_total_months(period):
    try:
        start_str, end_str = period.split(' - ')
        start_date = datetime.strptime(start_str, '%b %Y')
        end_date = datetime.strptime(end_str, '%b %Y')
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1  # +1 to include the starting month
    except:
        return None
```

```python
#applying the function
df['Months'] = df['TimeStamp'].apply(calculate_total_months)
```
